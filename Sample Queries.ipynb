```python
%sql
select 
p.BusinessEntityID as CustomerID, 
p.PersonType as CustomerType, 
p.FirstName as CustomerFirstName, 
p.LastName as CustomerLastName, 
AccountNumber as CustomerAccountNumber
--,current_timestamp() as RightNow

from silver.awfolderbased.person_person p 
inner join silver.awfolderbased.sales_customer c
on p.BusinessEntityID = c.PersonID
```
